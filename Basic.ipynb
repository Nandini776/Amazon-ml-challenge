```python
import pandas as pd

source1 = pd.read_csv("6ab10eb3b23ba_student_resource/student_resource/dataset/train/train_source1.tsv", sep="\t", nrows=5)
source2 = pd.read_csv("6ab10eb3b23ba_student_resource/student_resource/dataset/train/train_source2.tsv", sep="\t", nrows=5)
source3 = pd.read_csv("6ab10eb3b23ba_student_resource/student_resource/dataset/train/train_source3.tsv", sep="\t", nrows=5)
group = pd.read_csv("6ab10eb3b23ba_student_resource/student_resource/dataset/train/train_ground_truth.tsv", sep="\t", nrows=10)

print(source1.columns.tolist())
print(source2.columns.tolist())
print(source3.columns.tolist())
print(group.columns.tolist())
```

```python
import re
import unicodedata

def clean_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text)
    
    # Normalize Unicode characters
    text = unicodedata.normalize("NFKC", text)
    
    # Convert English characters to lowercase
    text = text.lower()
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)
    
    # Remove leading/trailing spaces
    text = text.strip()
    
    return text
```

```python
for df in [source1, source2, source3]:
    df["business_name_clean"] = df["business_name"].apply(clean_text)
    df["business_address_clean"] = df["business_address"].apply(clean_text)
    df["country_clean"] = df["country"].apply(clean_text)
```
